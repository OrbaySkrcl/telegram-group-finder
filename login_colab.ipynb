```python
import nest_asyncio
nest_asyncio.apply()

from telethon.sync import TelegramClient
from telethon.sessions import StringSession

api_id = int(input("api_id: ").strip())
api_hash = input("api_hash: ").strip()

with TelegramClient(StringSession(), api_id, api_hash) as client:
    me = client.get_me()
    print("\nGiris basarili:", me.username or me.first_name)
    print("\n" + "=" * 60)
    print("Asagidaki satirin TAMAMINI kopyala:")
    print("=" * 60)
    print(client.session.save())
    print("=" * 60)
```

## Sırada ne var?

Kopyaladığın metni Railway'de şuraya gireceksin:

| Değişken | Değer |
|---|---|
| `TG_API_ID` | yukarıda girdiğin api_id |
| `TG_API_HASH` | yukarıda girdiğin api_hash |
| `TG_SESSION` | az önce kopyaladığın uzun metin |
| `DB_PATH` | `/data/tgfinder.db` |

Ayrıca Railway'de servise **Volume** ekleyip mount path'ini `/data` yapmayı unutma —
yoksa her güncellemede topladığın tüm veri silinir.
